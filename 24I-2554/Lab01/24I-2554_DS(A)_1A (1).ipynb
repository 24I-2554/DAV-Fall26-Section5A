```python
import pandas as pd
import numpy as np
import sqlite3

df = pd.read_csv('adult.csv')
df.head()
```

## Task 1 

```python
df.info()
print('Before replacing ?:')
print(df.isnull().sum())

df.replace('?', np.nan, inplace=True)

print('After replacing ?:')
print(df.isnull().sum())
```

## Task 2 

```python
missing = pd.DataFrame({'Missing Count': df.isnull().sum(), 'Missing Percentage': (df.isnull().sum() / len(df)) * 100})
missing.sort_values('Missing Count', ascending=False)
```

## Task 3 

Missing values are categorical, so they are filled with Unknown. No numeric column has missing values.

```python
df['workclass'].fillna('Unknown', inplace=True)
df['occupation'].fillna('Unknown', inplace=True)
df['native.country'].fillna('Unknown', inplace=True)
print(df.isnull().sum())
```

## Task 4
Only exact duplicates are removed. Duplicates without the income column are kept because different people can have the same attributes but different income labels.

```python
print('Exact duplicates:', df.duplicated().sum())
print('Duplicates without income:', df.drop(columns=['income']).duplicated().sum())

df = df.drop_duplicates()
print('Shape after removing exact duplicates:', df.shape)
```

## Task 5 

```python
print('Education:')
print(df['education'].unique())
print('\nMarital Status:')
print(df['marital.status'].unique())
print('\nNative Country:')
print(df['native.country'].unique())

for column in df.select_dtypes(include='object').columns:
    df[column] = df[column].str.strip()
```

## Task 6 

```python
print('Numeric columns:')
display(df.describe())
print('Categorical columns:')
display(df.describe(include='object'))
print('Mean and median:')
print(df[['age', 'hours.per.week']].agg(['mean', 'median']))
```

Mean age is slightly higher than median age. Mean hours per week and median hours per week are very close.

## Task 7 

```python
print('Most common occupation:')
print(df['occupation'].value_counts().head(1))
print('\nSex percentage:')
print(df['sex'].value_counts(normalize=True) * 100)
print('\nIncome percentage:')
print(df['income'].value_counts(normalize=True) * 100)
```

The income classes are imbalanced because <=50K has many more records than >50K.

## Task 8 

```python
education_check = df.groupby('education')['education.num'].unique()
print(education_check)

inconsistent = education_check[education_check.apply(len) > 1]
print('\nInconsistencies:')
print(inconsistent)

if len(inconsistent) == 0:
    print('No inconsistencies found.')
```

## Task 9 

```python
raw_df = pd.read_csv('adult.csv')
connection = sqlite3.connect(':memory:')
raw_df.to_sql('adult_income', connection, index=False, if_exists='replace')

sql_df = pd.read_sql_query('SELECT * FROM adult_income', connection)
age_over_30 = pd.read_sql_query('SELECT * FROM adult_income WHERE age > 30', connection)

print('Original CSV shape:', raw_df.shape)
print('SQL SELECT * shape:', sql_df.shape)
print('Age > 30 shape:', age_over_30.shape)
display(age_over_30.head())
connection.close()
```

```python
df.to_csv('adult_cleaned.csv', index=False)
print('adult_cleaned.csv saved successfully')
```
